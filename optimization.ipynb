```python
from bird_and_swarm import swarm
import math
```

```python
# define function

function_type = 'parabola'
def optimization_function(x, y):
    return (x**2 + y**2)

new_swarm = swarm(size=100, scorer = optimization_function, bounds=100)
```

```python
swarm.plot_birds(new_swarm, filename=f'data/{function_type}/frame0')

iterations = 40

for i in range(iterations):
    swarm.update_all(new_swarm)
    # print(f"Iteration: {i+1}, Best Score {new_swarm.best_score:.6f} at ({new_swarm.x_group_best:.4f}, {new_swarm.y_group_best:.4f})")
    swarm.plot_birds(new_swarm, filename=f'data/{function_type}/frame{i+1}')
```

```python
import imageio.v2 as imageio
images = []
filenames = []

for i in range(iterations + 1):
    filenames.append(f'data/{function_type}/frame{i}.png')

for filename in filenames:
    images.append(imageio.imread(filename))
imageio.mimsave(f'data/{function_type}/animation.gif', images, duration=0.5)
```
