```python
from bird_and_swarm import swarm
import math
```

```python
# define function
def optimization_function(x, y):
    return x**2 + y**2 + math.sin(x * y)

new_swarm = swarm(size=1000, scorer = optimization_function)
```

```python
for i in range(20):
    swarm.update_all(new_swarm)
    print(f"Iteration: {i+1}, Best Score {new_swarm.best_score:.4f} at ({new_swarm.x_group_best:.4f}, {new_swarm.y_group_best:.4f})")
```
